# Basic Imports

```python
import pandas as pd
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
```

# Survival rate by class AND sex

```python
df.groupby(["Pclass" , "Sex"])["Survived"].mean()
```

# Reset index to get a flat DataFrame instead of a multi-index

```python
result = df.groupby(["Pclass" , "Sex"])["Survived"].mean().reset_index()
print(result)
```

Always call .reset_index() after a groupby if you want a clean flat DataFrame to work with further.
