# Mean

The average. Sum of all values divided by count.

```python
import numpy as np
import pandas as pd

salaries = np.array([35000, 42000, 38000, 45000, 250000])

mean = np.mean(salaries)
print(f"Mean: {mean:,.0f}")   # 82,000
```

The mean is pulled toward extreme values. One outlier (250,000) pushed the mean to 82,000 — which doesn’t represent any actual salary in this dataset. This is the most important limitation of the mean.
