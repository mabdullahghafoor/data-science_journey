# Basic Imports

```python
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
import seaborn as sns

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)
```

# Question: which numeric columns are correlated with survival?


```python
numeric_col = ["Survived" , "Pclass" , "Age" , "SibSp" , "Parch" , "Fare"]
corr = df[numeric_col].corr()


fig, ax = plt.subplots(figsize=(8,6))

sns.heatmap(corr,
            annot=True,          # show numbers in cells
            fmt=".2f",           # 2 decimal places
            cmap="coolwarm",     # red = positive, blue = negative
            center=0,            # center the colormap at 0
            square=True,
            linewidths=0.5,
            ax=ax)


ax.set_title("Fare is positively correlated with survival;\nclass is negatively correlated",
             fontsize = 14 , fontweight="bold")

plt.tight_layout()
plt.show()

```

The heatmap is almost always the first chart interviewers ask about. Know that values close to 1 mean strong positive relationship, close to -1 mean strong negative relationship, and close to 0 mean no linear relationship.
