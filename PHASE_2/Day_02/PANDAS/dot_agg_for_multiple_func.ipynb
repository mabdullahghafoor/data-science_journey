Instead of calling one function at a time, .agg() lets you apply several simultaneously.

# Basic Imports


```python
import pandas as pd
url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"
df = pd.read_csv(url)
```

# Multiple functions on one column

```python
df.groupby("Pclass")["Fare"].agg(["mean", "min", "max", "count"])
```

# Different functions on different columns

```python
df.groupby("Pclass").agg(
    avg_age = ("Age" , "mean"),
    avg_fare = ("Fare" , "mean"),
    survival_rate = ("Survived" , "mean"),
    total_passengers = ("PassengerId" , "count")
)
```

The named aggregation syntax new_col_name=("source_col", "function") is the cleanest way to write this — your output columns get proper names.
