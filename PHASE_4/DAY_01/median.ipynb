# Median

The middle value when sorted. If even count, average of the two middle values.

```python
import numpy as np
import pandas as pd

salaries = np.array([35000, 42000, 38000, 45000, 250000])
```

```python
median = np.median(salaries)
print(f"Median: {median:,.0f}")   # 42,000
```

The median ignores outliers entirely. 42,000 actually represents what a typical person earns in this dataset.
