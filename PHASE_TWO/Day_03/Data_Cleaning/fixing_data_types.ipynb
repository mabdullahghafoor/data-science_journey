Wrong data types are silent killers — your aggregations give wrong results and you don't know why.

# Basic Imports

```python
import pandas as pd
import numpy as np

url = "https://raw.githubusercontent.com/datasciencedojo/datasets/master/titanic.csv"

df = pd.read_csv(url)
```

# Check current types

```python
df.dtypes
```

# --- Convert to numeric ---

```python
df["Age"] = df["Age"].astype(int)       # float → int (only if no nulls remain)
df["Fare"] = df["Fare"].astype(float)
```

When a numeric column has dirty values mixed in (e.g. "35abc")


astype() will crash — use pd.to_numeric with errors="coerce"

```python
df["Age"] = pd.to_numeric(df["Age"] , errors="coerce")
```

"coerce" turns unparseable values into NaN instead of crashing


You can then handle those NaNs with fillna()

# --- Convert to datetime ---

```python
df["date"] = pd.to_datetime(df["date"])
df["date"] = pd.datetime(df["date"] , format="%Y-%m-%d")
```

# Once it's datetime, you can extract parts

```python
df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month
df["day"] = df["date"].dt.day
```

# --- Convert to category ---

Saves memory and signals to libraries that this is categorical

```python
df["Pclass"] = df["Pclass"].astype("Category")
df["Survived"] = df["Survived"].astype("Survived")
```
