# Mode

The most frequently occurring value.

```python
import numpy as np
import pandas as pd
import scipy


salaries = np.array([35000, 42000, 38000, 45000, 250000])
```

```python
from scipy import stats

data = np.array([2, 3, 3, 4, 4, 4, 5, 5])
mode = stats.mode(data)
print(f"Mode: {mode.mode[0]}")   # 4
```

Mode is most useful for categorical data — the most common city, the most purchased product, the most common age group.
